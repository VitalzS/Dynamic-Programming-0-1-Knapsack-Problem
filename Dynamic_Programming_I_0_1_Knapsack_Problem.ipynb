```python
import ast

# Open and read the data from 1.txt
with open("10 (1).txt", "r") as file:
    raw_data = file.read().strip()

# Replace curly braces with square brackets to parse them as lists instead of sets
# This prevents the 'unhashable type: set' error
formatted_data = raw_data.replace('{', '[').replace('}', ']')

# Convert the formatted text into Python lists and numbers
data = ast.literal_eval(formatted_data)

# Get the knapsack capacity and the list of bricks
capacity = data[0]
bricks = data[1]

# Create a list to store the maximum profit for each capacity
profit = [0] * (capacity + 1)

# Go through each brick
for brick in bricks:
    value = brick[0]
    weight = brick[1]

    # Start from the largest capacity and work backward
    for current_capacity in range(capacity, weight - 1, -1):

        # Calculate the profit if we take this brick
        new_profit = profit[current_capacity - weight] + value

        # Keep whichever profit is larger
        if new_profit > profit[current_capacity]:
            profit[current_capacity] = new_profit

# Print the maximum possible profit
print("Maximum profit:", profit[capacity])
```
